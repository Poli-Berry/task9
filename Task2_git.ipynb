```python
"""Count string elements in a mixed-type list."""

data = [
    42,
    "hello",
    3.14,
    "world",
    True,
    "python",
    None,
    [1, 2],
    "data",
]

string_count = 0

for element in data:
    if type(element) is str:
        string_count += 1

print(f"Total string elements found: {string_count}")
```
